Визуализация распределение значений признаков Survived, Pclass, Age, Sex. Parch с помощью sns.countplot

```python
import pandas as pd
import matplotlib.pyplot as plt

# Считываем датасет из файла train.csv
data = pd.read_csv('train.csv')

# Визуализируем распределение значений признака Survived
plt.figure(figsize=(10, 5))
sns.countplot(x='Survived', data=data)
plt.title('Распределение значений признака Survived')
plt.show()

# Визуализируем распределение значений признака Pclass
plt.figure(figsize=(10, 5))
sns.countplot(x='Pclass', data=data)
plt.title('Распределение значений признака Pclass')
plt.show()

# Визуализируем распределение значений признака Age
plt.figure(figsize=(10, 5))
sns.histplot(data=data, x='Age', kde=True)
plt.title('Распределение значений признака Age')
plt.show()

# Визуализируем распределение значений признака Sex
plt.figure(figsize=(10, 5))
sns.countplot(x='Sex', data=data)
plt.title('Распределение значений признака Sex')
plt.show()

# Визуализируем распределение значений признака Parch
plt.figure(figsize=(10, 5))
sns.countplot(x='Parch', data=data)
plt.title('Распределение значений признака Parch')
plt.show()
```

Визуализация распределение значений признаков Survived, Pclass, Age, Sex. Parch с помощью ax.hist

```python
import pandas as pd
import matplotlib.pyplot as plt

# Считываем датасет из файла train.csv
data = pd.read_csv('train.csv')

# Визуализируем распределение значений признака Survived
fig, ax = plt.subplots(figsize =(10, 5))
ax.hist(x='Survived', data=data)
plt.xlabel("Survived")
plt.ylabel("count")
plt.title('Распределение значений признака Survived')
plt.figure(figsize=(20, 5))
plt.show()

# Визуализируем распределение значений признака Pclass
fig, ax = plt.subplots(figsize =(10, 5))
ax.hist(x='Pclass', data=data)
plt.xlabel("Pclass")
plt.ylabel("count")
plt.title('Распределение значений признака Pclass')
plt.show()


# Визуализируем распределение значений признака Age
fig, ax = plt.subplots(figsize =(10, 5))
ax.hist(x='Age', data=data)
plt.xlabel("Age")
plt.ylabel("count")
plt.title('Распределение значений признака Age')
plt.show()


# Визуализируем распределение значений признака Sex
fig, ax = plt.subplots(figsize =(10, 5))
ax.hist(x='Sex', data=data)
plt.xlabel("Sex")
plt.ylabel("count")
plt.title('Распределение значений признака Sex')
plt.show()


# Визуализируем распределение значений признака Parch
fig, ax = plt.subplots(figsize =(10, 5))
ax.hist(x='Parch', data=data)
plt.xlabel("Parch")
plt.ylabel("count")
plt.title('Распределение значений признака Parch')
plt.show()

```

Построение графика типа boxplot для столбца Age

```python
plt.figure(figsize=(30, 5))
sns.boxplot(x='Age', data=data, showmeans=True, meanline=True)
plt.title('Распределение значений признака Age')
plt.show()
```

**Интерпретация полученного графика:
После анализа BOX Plot можно сделать вывод о том, что большинство пассадиров имеют возраст от 20 до 38 лет. Есть несколько выбросов великовозрастных людей, котоые не вощли в Box plot.**

Построение график типа pie chart для переменных Survived, Pclass:

```python
import pandas as pd
import matplotlib.pyplot as plt

# Считываем датасет из файла train.csv
data = pd.read_csv('train.csv')

# Построение графика типа pie chart для переменной Survived
plt.figure(figsize=(10, 5))
plt.pie(data['Survived'].value_counts(),
                                      labels=['Не выжил', 'Выжил'],
                                     autopct='%1.1f%%')
plt.title('Распределение значений признака Survived')
plt.show()
```

```python
# Построение графика типа pie chart для переменной Pclass
plt.figure(figsize=(10, 5))
plt.pie(data['Pclass'].value_counts(),
                                   labels=['Первый класс', 'Второй класс', 'Третий класс'],
                                 autopct='%1.1f%%')
plt.title('Распределение значений признака Pclass')
plt.show()
```

Построение графика типа pairplot для всех числовых переменных датасета:


```python
sns.pairplot(data=data, diag_kind='kde', kind='scatter') # Построение графика типа pairplot для всех числовых переменных датасета:
plt.show()
```

Построение интерактивного sunburst plot (визуализация иерархических данных) с помощью plotly.На первом уровне иерархии - количество пассажиров в каждом из классов, а на втором количество женщин/мужчин в этом классе

```python
import pandas as pd
import plotly.express as px

# Считываем данные из файла train.csv
df = pd.read_csv('train.csv')

# Создаём sunburst plot с помощью plotly
fig = px.sunburst(df, path=['Pclass', 'Sex'], values='Survived')

# Добавляем заголовок к графику
fig.update_layout({'title': 'Выживаемость пассажиров на Титанике по классам и полу'})

# Отображаем график
fig.show()
```
