Цена и  аренда вряд ли больше +2147483647 рублей. Можно изменить тип до int32

Ванные комнаты, количество этажей, количество парковочных мест можно уложить в диапозон int8
там где тип object, его можно заменить на category, так как ограниченный диапозон значений - только "да" или "нет"

```python
data2['price'] = data2['price'].astype('int32')#Заменяем все исходя из коммента выше
data2['area'] = data2['area'].astype('int32')
data2['bedrooms'] = data2['bedrooms'].astype('int8')
data2['bathrooms'] = data2['bathrooms'].astype('int8')
data2['stories'] = data2['stories'].astype('int8')
data2['mainroad'] = data2['mainroad'].astype('category')
data2['guestroom'] = data2['guestroom'].astype('category')
data2['basement'] = data2['basement'].astype('category')
data2['hotwaterheating'] = data2['hotwaterheating'].astype('category')
data2['airconditioning'] = data2['airconditioning'].astype('category')
data2['parking'] = data2['parking'].astype('int8')
data2['prefarea'] = data2['prefarea'].astype('category')
data2['furnishingstatus'] = data2['furnishingstatus'].astype('category')
```

```python
memory_after = data2.memory_usage(deep=True).sum() #Выводим количесвто потребляемой памяти после замены типа
memory_after
```

```python
memory_before/memory_after
```

Видно, что получилось получить выигрыш в 20 раз
